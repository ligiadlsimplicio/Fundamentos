```python
import pandas as pd

# 1. Criação das duas tabelas fictícias
df_alunos = pd.DataFrame(
    {"Identificador": [1, 2, 3], "Nome_Aluno": ["Ana Silva", "Beto Costa", "Carla Souza"]}
)

df_cursos = pd.DataFrame(
    {
        "Identificador_Aluno": [1, 2, 3],
        "Curso": ["Engenharia de Dados", "Ciência de Dados", "Análise de Dados"],
    }
)

# 2. Junção das tabelas usando .merge()
df_final = df_alunos.merge(
    df_cursos, left_on="Identificador", right_on="Identificador_Aluno", how="inner"
)

# 3. Exibição da tabela unificada
print(df_final)
```
