```python
from pathlib import Path
import pandas as pd

# Carrega os dados do arquivo Parquet
caminho = (
    Path("artigos.parquet")
    if Path("artigos.parquet").exists()
    else Path("../data/exemplo/artigos.parquet")
)
df = pd.read_parquet(caminho)

# Identifica a coluna correta de autor/pesquisador
col_autor = "pesquisador" if "pesquisador" in df.columns else "nome_completo"

# Agrupamento e ordenação decrescente
ranking = df.groupby(col_autor).size().sort_values(ascending=False)
print("--- RANKING DE PUBLICAÇÕES POR PESQUISADOR ---")
print(ranking)
```

**Análise de Ranking por Pesquisador:**
* O agrupamento identifica os docentes mais produtivos do conjunto de dados.
* A liderança do ranking concentra a maior volumetria de publicações, permitindo comparar o volume de entregas entre os autores analisados.

```python
# Agrupamento por ano e ordenação cronológica
evolucao_temporal = df.groupby("ano").size().sort_index()

print("--- EVOLUÇÃO HISTÓRICA DE PUBLICAÇÕES POR ANO ---")
print(evolucao_temporal)
```

**Análise de Evolução Temporal:**
* O agrupamento por ano demonstra a oscilação histórica na produção acadêmica.
* A série temporal permite identificar picos de produtividade, além de detectar anos com baixa emissão ou ausência de publicações.
