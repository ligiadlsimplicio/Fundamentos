```python
from datetime import datetime
import pandas as pd

# Dicionário com dados fictícios de funcionários
dados_funcionarios = {
    'Nome': ['Ana Silva', 'Bruno Costa', 'Carla Souza', 'Diego Lima'],
    'Cargo': [
        'Analista de Dados',
        'Desenvolvedor Backend',
        'Gerente de Projetos',
        'Designer UX',
    ],
    'Idade': [28, 35, 42, 30],
    'Salário': [5500.00, 7200.00, 11000.00, 6000.00],
    'Tempo de Empresa': [3, 5, 10, 2],
}

# Criando o DataFrame
df = pd.DataFrame(dados_funcionarios)

# Exibindo o DataFrame
df
```

```python
# Selecionando uma coluna (retorna uma Series)
coluna_nome = df['Nome']
coluna_nome
```

```python
# Selecionando duas colunas (retorna um DataFrame)
subconjunto = df[['Nome', 'Salário']]
subconjunto
```

```python
# Obtendo o ano atual dinamicamente e calculando o ano de entrada
ano_atual = datetime.now().year
df['Ano de Entrada'] = ano_atual - df['Tempo de Empresa']

# Exibindo a tabela atualizada
df
```

```python
# Criando uma Series a partir da coluna 'Salário' usando o 'Nome' como índice
salarios_series = pd.Series(
    data=df['Salário'].values, index=df['Nome'], name='Salário Atual'
)

salarios_series
```
